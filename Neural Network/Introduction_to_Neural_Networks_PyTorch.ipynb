Neural Networks are computational models that mimic the complex functions of the human brain. The neural networks consist of interconnected nodes or neurons that process and learn from data, enabling tasks such as pattern recognition and decision making in machine learning. This lab introduces PyTorch and demonstrates how NN models are implemented.

```python
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
```

# Neuron with Sigmoid activation

```python
# Define the dataset
X_train = np.array([0., 1, 2, 3, 4, 5], dtype=np.float32).reshape(-1,1)
y_train = np.array([0,  0, 0, 1, 1, 1], dtype=np.float32).reshape(-1,1)

# Convert to PyTorch tensors
X_train_t = torch.tensor(X_train)
y_train_t = torch.tensor(y_train)
```

```python
# Build the model (1 neuron + sigmoid)
model = nn.Sequential(
    nn.Linear(1, 1),
    nn.Sigmoid()
)
print(model)
```

```python
# Setting the weight and bias manually for demonstration
with torch.no_grad():
    model[0].weight.copy_(torch.tensor([[2.0]]))
    model[0].bias.copy_(torch.tensor([-4.5]))

print("Weight:", model[0].weight.item())
print("Bias:", model[0].bias.item())
```

```python
def sigmoid_np(z):
    return 1 / (1 + np.exp(-z))
```

#### Comparing the prediction with numpy prediction vs PyTorch prediction

```python
# PyTorch prediction
model.eval()
with torch.no_grad():
    a1_pt = model(torch.tensor([[0.0]]))
print("PyTorch Prediction for X=0:", a1_pt.item())

# Numpy prediction
z = np.dot(np.array([[0.0]]), np.array([[2.0]])) + (-4.5)
a1_np = sigmoid_np(z)
print("NumPy Prediction for X=0:", a1_np[0][0])
```
