Model 2 generally achieves significantly higher accuracy than Model 1

Adding another hidden layer gives the network a hierarchical composition of features, so it can model interactions between elevation, distance, soil, and wilderness indicators more effectively than a single hidden layer. That extra composition is the practical benefit of depth in a feedforward ANN

Hidden layer 1: `(54 × 64) + 64 = 3,520`  
Hidden layer 2: `(64 × 32) + 32 = 2,080`  
Output layer: `(32 × 7) + 7 = 231`  

Total trainable parameters: `3,520 + 2,080 + 231 = 5,831`

Combine those simple features into complex, non-linear representations. Since forest cover types aren't separated by straight lines in a 54-dimensional space, the extra "curviness" provided by more ReLU activations helps the model fit the decision boundaries

Dropout randomly zeros activations during training, which reduces co-adaptation and usually improves generalization when the model starts to overfit. BatchNormalization keeps activations better scaled, which often makes optimization faster and more stable
